```python
import pandas as pd
from matplotlib import pyplot as plt
df=pd.read_csv("C:\\Users\\padma\\OneDrive\\heat_dissipation_with_heat_loss.csv")
x=df[['Material_Thermal_Conductivity','Surface_Area','Contact_Pressure','Heat_Source_Power','Time','Operating_Temperature','Thickness','Ambient_Temperature']]
y=df[['Heat_Loss']]
```

```python
import pandas as pd
from matplotlib import pyplot as plt
df=pd.read_excel("C:\\Users\\padma\\Downloads\\Book12.xlsx")
```

```python

labels = df['Assembly_Type']
sizes = df['Heat_Loss']

plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.axis('equal')  # Optional, but recommended
plt.show()
```

```python
avg_loss = df.groupby('Assembly_Type')['Heat_Loss'].mean()

# Plot line chart
plt.plot(avg_loss.index, avg_loss.values, marker='o', linestyle='-')
plt.xlabel('Assembly Type')
plt.ylabel('Average Heat Loss')
plt.title('Average Heat Loss by Assembly Type')
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```

```python
plt.bar(df['Assembly_Type'],df['Heat_Loss'],)
plt.title('Average Heat Loss by Assembly Type')
plt.xlabel('Assembly Type')
plt.ylabel('Average Heat Loss')
plt.grid(axis='y')
plt.show()
```
