```python
import pandas as pd
import matplotlib.pyplot as plt
# data
df=pd.read_csv("machine_failure_dataset_1.csv")
# Group data by Machine_Type and sum the failures
failure_data = df.groupby('Machine_Type')['failures'].sum()
# Bar Chart
plt.figure(figsize=(8, 5))
failure_data.plot(kind='bar', color='skyblue')
plt.title('Total Failures per Machine Type - Bar Chart')
plt.xlabel('Machine Type')
plt.ylabel('Failures')
plt.grid(axis='y')
plt.tight_layout()
plt.show()

```

```python

# Line Chart
plt.figure(figsize=(8, 5))
failure_data.plot(kind='line', marker='o', color='green')
plt.title('Total Failures per Machine Type - Line Chart')
plt.xlabel('Machine Type')
plt.ylabel('Failures')
plt.grid(True)
plt.tight_layout()
plt.show()

```

```python
# Pie Chart
plt.figure(figsize=(7, 7))
failure_data.plot(kind='pie', autopct='%1.1f%%', startangle=140, colormap='Pastel1')
plt.title('Failure Distribution by Machine Type - Pie Chart')
plt.ylabel('')
plt.tight_layout()
plt.show()
```
